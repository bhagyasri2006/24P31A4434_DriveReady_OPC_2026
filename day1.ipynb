```python
from langchain_groq import ChatGroq
print("LangChain Groq installed successfully!")
```

```python

from langchain_groq import ChatGroq

llm = ChatGroq(
    model="openai/gpt-oss-20b",
    temperature=0
)

print("Groq LLM connected successfully!")
```

```python

SYSTEM_PROMPT = """
You are an AI Customer Complaint Classifier.

Analyze the customer's complaint and identify:

1. Category: Payment, Delivery, Product, Refund, or Other.
2. Priority: Low, Medium, or High.
3. Reason: A short explanation for your classification.

Return the result in a clear, structured format.
"""

def classify_complaint(complaint):
    response = llm.invoke([
        ("system", SYSTEM_PROMPT),
        ("human", complaint)
    ])
    return response.content
```

```python

complaint = "I paid for my order, but the amount was deducted twice."

result = classify_complaint(complaint)

print(result)
```

```python

test_complaints = [
    "My order has not arrived even after 10 days.",
    "The product I received is damaged.",
    "I requested a refund, but I haven't received my money.",
    "The website keeps crashing whenever I try to place an order."
]

for complaint in test_complaints:
    print("\nComplaint:", complaint)
    print(classify_complaint(complaint))
    print("-" * 50)
```

```python

def run_classifier():
    complaint = input("Enter customer complaint: ")

    if not complaint.strip():
        print("Please enter a valid complaint.")
        return

    result = classify_complaint(complaint)

    print("\n--- Complaint Analysis ---")
    print(result)


run_classifier()
```
